```python
import pandas as pd

df = pd.read_csv("zomato.csv")
df.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['rate'] = df['rate'].astype(str)
df['rate'] = df['rate'].str.strip()
df['rate'] = df['rate'].str.replace('/5', '', regex=False)
df['rate'] = df['rate'].replace(['NEW', '-', 'nan', 'NaN', ''], None)
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
df = df.dropna(subset=['rate'])
```

```python
sns.histplot(df['rate'], bins=20)
plt.title("Ratings Distribution")
plt.show()
```

Most restaurants have ratings between 3.5–4.2.  
Very few restaurants have low ratings.  
Highly rated restaurants are limited.

```python
df['location'].value_counts().head(10).plot(kind='bar')
plt.title("Top Locations")
plt.show()
```

Some locations have more restaurants than others.  
This shows higher demand in those areas.

```python
sns.scatterplot(x='approx_cost(for two people)', y='rate', data=df)
plt.title("Cost vs Rating")
plt.show()
```

Higher cost does not always mean better ratings.

```python
df['cuisines'].value_counts().head(10).plot(kind='bar')
plt.title("Top Cuisines")
plt.show()
```

Some cuisines are more popular than others.

**Final Insights**

Most restaurants have ratings between 3.5–4.2  
Location plays an important role  
High ratings are rare  
Price does not guarantee better rating
